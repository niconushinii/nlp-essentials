```python
from collections import Counter

def count_words(corpus: str) -> Counter:
    fin = open(corpus)
    words = fin.read().split()
    return Counter(words)
```


    L1: Import the Counter class, a special type of a dictionary, from the collections package.
    L3: Use typing to indicate the parameter type (str) and the return type (Counter).
    L4: open() the corpus file.
    L5: read() the contents of the file as a string, split() it into a list of words, and store them in words.
    L6: Count the occurrences of each word and return the results as a Counter.


```python
corpus = 'dat/emory-wiki.txt'
counts = count_words(corpus)

n_tokens = sum(counts.values())
n_types = len(counts)

print(f'# of word tokens: {n_tokens}')
print(f'# of word types: {n_types}')
```


    L1: The corpus can be found dat/emory-wiki.txt.
    L4: Save the total number of word tokens in the corpus, which is the sum() of counts values.
    L5: Save the number of unique word types in the corpus, which is the len() of counts.
    L7-8: Print the value using the formatted string literals.
